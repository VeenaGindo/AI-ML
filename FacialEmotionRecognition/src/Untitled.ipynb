```python
from PIL import Image
import matplotlib.pyplot as plt
import cv2

def resize_image(image_path, target_size=(48, 48)):
  """Resizes an image to the target size."""
  try:
    img = Image.open(image_path)
    img = img.convert('L')
    img = img.resize(target_size, Image.BILINEAR)  # Or use Image.NEAREST for faster, but potentially less accurate resizing
    return img
  except FileNotFoundError:
    print(f"Error: Image not found at {image_path}")
    return None
  except Exception as e:
    print(f"Error processing image: {e}")
    return None

# Example usage:
image_path = "C:\\Users\\veena\\CloudxLabs\\Kaggle\\Datasets\\FacialEmotionRecognition\\images\\3\\Neutral.jpg"
resized_image = resize_image(image_path)

if resized_image:
  # Now you can use resized_image with your model
  # ...
#   plt.figure(figsize=(20, 20))
  resized_image.save("C:\\Users\\veena\\CloudxLabs\\Kaggle\\Datasets\\Resized\\resized_image1.jpg") # optional: save the resized image
#   plt.imshow(cv2.cvtColor(resized_image, cv2.COLOR_BGR2RGB))
#   plt.imshow(resized_image)
#   plt.axis('off')
#   plt.show()
```
